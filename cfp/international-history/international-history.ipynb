# Investigating the Use of Digital Methods in Diplomatic and International History

## A special issue for the *Journal of Digital History*

Proposal by Laurence Badel, Louis Clerc and Myriam Piguet.

In recent years, the history of international relations and diplomacy has shifted focus. Sweeping away from the traditional realistic perspective, the field has given more space to the multiplicity of practices, actors, and types of agency, moving toward a more socio-historic approach of international relations <cite id="b9768"><a href="#zotero%7C8918853%2FEWFNQTIA">(Kott, 2011)</a></cite>. In the past 20 years, many studies have thus investigated previously underexplored spaces of international relations: international organization secretariats (see for instance <cite id="46fe4"><a href="#zotero%7C8918853%2FVJC4LKEJ">(Gram-Skjoldager et al., 2020; Reinalda, 2020)</a></cite>), non-governmental organizations and international associations (see for instance <cite id="97f2a"><a href="#zotero%7C8918853%2FHRM3DQ7A">(Gradskova, 2021; Guieu, 2008)</a></cite>), and even the office space of international settings such as European organizations (see for instance the [BUREU research project](https://www.uni.lu/c2dh-en/articles/bureu-a-new-transnational-research-project-crossing-the-border-between-architectural-history-and-european-integration-history/), led by Andreas Fickers and Martin Kohlrausch, on the nature of EU office buildings and their ties with managerial principles). At the same time, historians of international relations and diplomacy have sought the human stories beyond the top diplomats, uncovering the stories of previously forgotten actors (for instance, the [INVISIHIST research project](https://www.universiteitleiden.nl/en/research/research-projects/humanities/the-invisible-history-of-the-united-nations-and-the-global-south-invisihist#tab-1) led by Alanna O'Malley on the invisible histories of the UN and the historical agency of the Global South), including silenced networks <cite id="2a0ad"><a href="#zotero%7C8918853%2FHAW8F9QB">(Avaro &#38; Gautier-Morin, 2026)</a></cite> and auxiliary roles <cite id="56a09"><a href="#zotero%7C8918853%2FPMMHT52Y">(Badel, 2024)</a></cite>.

With this shift, the field has thoroughly transformed. At present, the historical community considers international relations not only as the balance between states' power, but also as a space where interactions and networks collide. This perspective creates infinite possibilities for novel interpretation, giving space to all the entities found in diplomacy: great powers, perceived middle-sized and small states, international organizations, civil society, and everything within them and in between. Furthermore, the integration of concepts such as decolonization, intersectional approach, and/or from below perspectives has also allowed the field to successfully reinvent itself.

This shift, however, equally owes to the changes in historical practices induced by the growth of digital methods. As recently stated by Clavert and Muller, digital methods have invaded most of the traditional motions and experiences of the historians, even for those of us that don't use digital tools on a daily basis <cite id="de05c"><a href="#zotero%7C8918853%2FSPVWPMRN">(Muller &#38; Clavert, 2025)</a></cite>. In the case of the history of international relations and of diplomacy, these practices have fundamentally changed our work methods: historians of international relations and of diplomacy currently face a growing number of accessible digitized archives (such as the now-famous [LONTAD project](https://libraryresources.unog.ch/lontad), Total Digital Access to the League of Nations Archives), have the possibility to uncover a dense web of networks through various digital tools (such as Nodegoat; see notably the work of Martin Grandjean, University of Lausanne). In the wider historical field, the use of digitally native archives has been discussed (see the [LUSTRE project](https://lustre-network.net/team/) led by Prof. Lise Jaillant) as well as the use of artificial intelligence and the ways in which Large Language Models could be ethically integrated into the historian's toolkit (see the [AI manifesto](https://www.uni.lu/c2dh-en/articles/ai-manifesto/) by the C²DH AI working group). This special issue aims to contribute to these debates from the point of view of Diplomatic and International Relations History.

In this context, this special issue of the *Journal of Digital History*, entitled *Investigating the Use of Digital Humanities in Diplomatic and International Relations History*, will uncover the multiple ways historians of international relations and of diplomacy have used digital tools into their research. The special issue will offer an in-depth review of the digital methods the field has integrated in recent years, providing a guide and suggesting potential responses to essential questions about the ethics and relevance of these methods.

The *Issue* will be divided into two core themes:
1. articles engaging with past experiences/innovations of the last ten years, and their outputs;
2. propositions for current in-design innovative digital humanities approaches that have the potential to transform the field even further.

These articles will have to be based on worked out examples and/or prototypes.

We are looking for proposals from historians of international relations and of diplomacy (17th–21st century) as well as digital humanists and engineers. Regarding the unique nature of the *Journal of Digital History*, we strongly encourage teams made of a blend of historians and engineers to apply and co-sign articles. Furthermore, this call welcomes proposals from non-permanent academic staff, including PhD researchers, postdocs, and research technicians.

## Information about the publication process at the JDH

**Before submitting a proposal, authors should note that the *Journal of Digital History* (JDH) employs an innovative three-layered digital format:**

1. **Narrative Layer:** the traditional academic text, as found in most journals, though exploring the many narrative possibilities that the web offers;
2. **Hermeneutic Layer:** highlighting the methodological implications of using digital tools, data and code;
3. **Code and Data Layer:** access to the original research data and any code used for analysis to allow readers to explore and/or replicate digital approaches.

Because of this unique set up, the publishing process at the *Journal of Digital History* differs from that of traditional academic journals. It includes the use of GitHub, Jupyter Notebook, and Zotero. Furthermore, the draft must be shared with the JDH editorial team in the form of a Jupyter Notebook and not in Word format. The JDH team, however, will provide a training seminar and guidance to adapt research into this digital format.

While code is not mandatory to publish with JDH, digital material is expected. Therefore, we strongly advise interested authors to apply as interdisciplinary duos or teams including digital humanists and/or developers. Furthermore, this method implies a longer publishing process than other academic journals.

To see this format in action, see for instance [this article](https://journalofdigitalhistory.org/en/article/BHmHNQKJaSWT). For detailed instructions on the publication process, see the JDH's [Guidelines for authors](https://journalofdigitalhistory.org/en/guidelines).
